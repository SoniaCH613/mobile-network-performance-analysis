    on trouve ces colonnes vide grace a la condition (nombre_manquantes==total_lignes) qui compare chaque valeur du tableau combres_manquante avec le nombre total de lignes qu'on a de base si c'est bien le cas donc on recupere la colonne et on l'ajoute a notre liste colonnes_vide

**)colonnes_partielles : pareil meme principe avec colonnes_vide juste que on a 2 condition 
    la premiere c'est pour voir si on est pas sur une colonne vide pour bien separés les colonnes vides et partiellement renseignés (parceque la deuxieme condition ne suffit pas vu que une colonne vide peut etre considére comme partiellement renseignés

16. Examinons les variables candidates à partir des résultats

```python
variables_candidates=[
    "operator",
    "protocole",
    "zone",
    "bitrate_dl",
    "bitrate_ul",
    "acess_duration",
    "loaded_in_less_5_secondes",
    "loaded_in_less_10_secondes",
    "quality_correct",
    "quality_perfect",
    "upload_ok",
    "date_start",
    "insee_com",
    "techno_start",
    "ping_result",
    "rsrp",
    "rsrq",
    "rsrp",
    "transfert_file_size"
    "Result"
]

variables_presentes = []

for colonne in variables_candidates:
    if colonne in donnees_completes.columns:
        variables_presentes.append(colonne)


variables_absentes = []

for colonne in variables_candidates:
    if colonne not in donnees_completes.columns:
        variables_absentes.append(colonne)



print("Variables candidates présentes : ",len(variables_presentes))
print("Variables candidates absentes : ",variables_absentes)
```

**) le but de cette étape : c'est pour savoir si les variable selectionnées correspondent bien aux colonnes qui existe dans notre dataframe et ca permet de detecter les fautes de frappes , Une variable qu'on pensait présente, mais qui n'existe pas dans le fichier , Une différence de nom entre notre liste et le DataFrame. 

```python
donnees_candidates=donnees_completes[variables_presentes]

bilan_variables=pd.DataFrame({
    "Type de données": donnees_candidates.dtypes.astype(str),
    "Valeur manquantes": donnees_candidates.isna().sum(),
    "Pourcentage manquant (%)" : (donnees_candidates.isna().mean() *100 ).round(2),
    "Valeurs distinctes " : donnees_candidates.nunique()
})

display(bilan_variables)
```

17. Vérifier les indicateurs  associés au different usages 

```python
indicateurs_a_verifier= {
    "DOWNLOAD" : ["bitrate_dl"],
    "UPLOAD" : ["bitrate_ul"], 
    "WEB" : [
        "acess_duration",
        "loaded_in_less_5_secondes",
        "loaded_in_less_10_secondes",
    ],
    "STREAM" : ["quality_correct", "quality_perfect"],
    "ULH" : ["upload_ok"]
}

for protocole, colonnes in indicateurs_a_verifier.items() : 
    mesures_protocole=donnees_completes[donnees_completes["protocole"]==protocole]

    bilan = pd.DataFrame({
        "Valeurs renseignées" : mesures_protocole[colonnes].count(),
        "Valeurs manquantes" : mesures_protocole[colonnes].isna().sum(),
        "Manquantes (%)" : (mesures_protocole[colonnes].isna().mean() * 100).round(2)
    })

    print(f"\n{protocole} _ {len(mesures_protocole)} observations")
    display(bilan)
```

## Vérification des indicateurs par protocole

**Objectif :** vérifier la disponibilité des indicateurs associés à chaque protocole et identifier les éventuelles valeurs manquantes.

### 1. Définir les indicateurs à vérifier

On crée un dictionnaire qui associe chaque protocole à ses indicateurs :
- Les **clés** représentent les protocoles.
- Les **valeurs** représentent les noms des colonnes (indicateurs) associés à chaque protocole.

### 2. Parcourir le dictionnaire

On utilise une boucle `for` avec `.items()` pour parcourir les paires clé-valeur du dictionnaire :
- `protocole` récupère le nom du protocole.
- `colonnes` récupère toute la liste des indicateurs associés à ce protocole.

La boucle effectue donc une itération par protocole.

### 3. Filtrer les observations

On sélectionne les lignes de `donnees_completes` dont la valeur dans la colonne `protocole` correspond au protocole de l'itération actuelle.

Pour cela :
- `donnees_completes["protocole"]` sélectionne la colonne des protocoles.
- `== protocole` compare chaque valeur avec le protocole actuel.
- Les crochets `[]` conservent uniquement les lignes qui respectent cette condition.

On récupère ainsi toutes les observations du protocole actuel dans `mesures_protocole`.

### 4. Construire le bilan des indicateurs

On sélectionne les indicateurs du protocole avec `mesures_protocole[colonnes]`, puis on crée un DataFrame récapitulatif contenant trois informations :

- `.count()` : compte les valeurs renseignées pour chaque indicateur.
- `.isna().sum()` : compte les valeurs manquantes.
- `.isna().mean() * 100` : calcule le pourcentage de valeurs manquantes.
- `.round(2)` : arrondit le pourcentage à deux décimales.

**Important :** le bilan est calculé sur toutes les observations du protocole actuel, et non sur chaque ligne individuellement.
