17. Vérifier les indicateurs  associés au different usages 

```python
indicateurs_a_verifier= {
    "DOWNLOAD" : ["bitrate_dl"],
    "UPLOAD" : ["bitrate_ul"], 
    "WEB" : [
        "acess_duration",
        "loaded_in_less_5_secondes",
        "loaded_in_less_10_secondes",
    ],
    "STREAM" : ["quality_correct", "quality_perfect"],
    "ULH" : ["upload_ok"]
}

for protocole, colonnes in indicateurs_a_verifier.items() : 
    mesures_protocole=donnees_completes[donnees_completes["protocole"]==protocole]

    bilan = pd.DataFrame({
        "Valeurs renseignées" : mesures_protocole[colonnes].count(),
        "Valeurs manquantes" : mesures_protocole[colonnes].isna().sum(),
        "Manquantes (%)" : (mesures_protocole[colonnes].isna().mean() * 100).round(2)
    })

    print(f"\n{protocole} _ {len(mesures_protocole)} observations")
    display(bilan)
```

**) pour indicateurs a verifier : c'est un dictionnaire dans les clés sont les valeurs sont 
    les indicateurs 


     Vérifier la disponibilité des indicateurs pour chaque protocole

 Vérification des indicateurs par protocole

**Objectif :** vérifier la disponibilité des indicateurs associés à chaque protocole et identifier les éventuelles valeurs manquantes.

1. Définir les indicateurs à vérifier

On crée un dictionnaire qui associe chaque protocole à ses indicateurs :
- Les **clés** représentent les protocoles.
- Les **valeurs** représentent les noms des colonnes (indicateurs) associés à chaque protocole.

 2. Parcourir le dictionnaire

On utilise une boucle `for` avec `.items()` pour parcourir les paires clé-valeur du dictionnaire :
- `protocole` récupère le nom du protocole.
- `colonnes` récupère toute la liste des indicateurs associés à ce protocole.

La boucle effectue donc une itération par protocole.

 3. Filtrer les observations

On sélectionne les lignes de `donnees_completes` dont la valeur dans la colonne `protocole` correspond au protocole de l'itération actuelle.

Pour cela :
- `donnees_completes["protocole"]` sélectionne la colonne des protocoles.
- `== protocole` compare chaque valeur avec le protocole actuel.
- Les crochets `[]` conservent uniquement les lignes qui respectent cette condition.

On récupère ainsi toutes les observations du protocole actuel dans `mesures_protocole`.

 4. Construire le bilan des indicateurs

On sélectionne les indicateurs du protocole avec `mesures_protocole[colonnes]`, puis on crée un DataFrame récapitulatif contenant trois informations :

- `.count()` : compte les valeurs renseignées pour chaque indicateur.
- `.isna().sum()` : compte les valeurs manquantes.
- `.isna().mean() * 100` : calcule le pourcentage de valeurs manquantes.
- `.round(2)` : arrondit le pourcentage à deux décimales.

**Important :** le bilan est calculé sur toutes les observations du protocole actuel, et non sur chaque ligne individuellement.

Analyse de certaines variables

20. Comparer les performances par operateur 


20.1 Débit descendant — DOWNLOAD

```python
# Sélectionner les observations du protocole DOWNLOAD
mesures_download = donnees_completes[
    donnees_completes["protocole"] == "DOWNLOAD"
]

# Calculer les statistiques du débit descendant par opérateur
stats_debit_dl = mesures_download.groupby("operator")["bitrate_dl"].describe()

# Afficher les résultats
display(stats_debit_dl.round(2))

#bitrate_dl: c'est le debit descendant autrement dit c'est la vitesse a laquelle le telephone recoit les donnees depuis internet
#objectif : déterminer quel opérateur présente les meilleurs débits descendants selon les mesures réalisées par l'ARCEP.
#Remarque : resultat : Orange possede le meilleur debit decsendant
```

20.2 Débit montant — UPLOAD

```python
mesures_upload=donnees_completes[donnees_completes["protocole"]=="UPLOAD"]
stats_debit_ul=mesures_upload.groupby("operator")["bitrate_ul"].describe()
display(stats_debit_ul.round(2))
#bitrate_ul : c'est le debit montant c_a_d la vitesse d'envoi des données 
#Remarque : resultat : Orange possede le meilleur debit montant 
```

20.3 Navigation Web — WEB

```python
mesures_web = donnees_completes[
    donnees_completes["protocole"] == "WEB"
]

stats_web = mesures_web.groupby("operator")[
    [
        "acess_duration",
        "loaded_in_less_5_secondes",
        "loaded_in_less_10_secondes"
    ]
].describe()

display(stats_web.round(2))
```

20.4 Streaming vidéo — STREAM

```python
# Sélectionner les observations du protocole STREAM
mesures_stream = donnees_completes[
    donnees_completes["protocole"] == "STREAM"
]
# Calculer les taux de qualité vidéo par opérateur
taux_video = mesures_stream.groupby("operator")[
    ["quality_correct", "quality_perfect"] 
    #ici les indicateur pour stream qui est la lecture de video en continue sans coupure
].mean() * 100 # ici les indicateurs ont été evalués par 0 et 1 donc en binaire et donc on calcule la moyenne de ces 0 et 1

# Afficher les résultats
display(taux_video.round(2))
# Conclusion :
# Orange obtient les meilleurs résultats pour les deux indicateurs
# de qualité vidéo, avec 93,76 % de qualité correcte
# et 92,89 % de qualité parfaite.
# Free obtient les taux les plus faibles.
```

20.5 Envoi de fichier complémentaire — ULH

```python
# Sélectionner les observations du protocole ULH
mesures_ulh = donnees_completes[
    donnees_completes["protocole"] == "ULH"
]
# Calculer le taux de réussite des envois par opérateur
taux_upload_ulh = mesures_ulh.groupby("operator")["upload_ok"].mean() * 100

# Afficher les résultats
display(taux_upload_ulh.round(2))

# Conclusion :
# Orange obtient le meilleur taux de réussite des tests ULH avec 92,69 %.
# Free présente le taux le plus faible avec 86,25 %.
# Les résultats montrent des différences de réussite entre les opérateurs.
```

22. Définir les colonnes nécessaires à la base PostgreSQL

```python
COLONNES_SOURCE = [
    "operator",
    "protocole",
    "zone",
    "date_start",
    "insee_com",
    "bitrate_dl",
    "bitrate_ul",
    "acess_duration",
    "loaded_in_less_5_secondes",
    "loaded_in_less_10_secondes",
    "quality_correct",
    "quality_perfect",
    "upload_ok"
]

colonnes_absentes = [
    colonne for colonne in COLONNES_SOURCE
    if colonne not in donnees_completes.columns
]

if colonnes_absentes:
    raise KeyError(f"Colonnes absentes du CSV : {colonnes_absentes}")

print("Toutes les colonnes prévues pour PostgreSQL sont présentes.")
```

```python
donnees_pour_base = donnees_completes.loc[
    donnees_completes["protocole"].isin(PROTOCOLES_RETENUS),
    COLONNES_SOURCE
].copy() 

print("Dimensions des données destinées à PostgreSQL :", donnees_pour_base.shape)
print("\nColonnes conservées :")
print(donnees_pour_base.columns.tolist())

display(donnees_pour_base.head())
```
