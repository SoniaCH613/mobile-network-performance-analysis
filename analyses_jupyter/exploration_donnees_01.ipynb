1. Importation des bibliothèques

```python
import pandas as pd
from pathlib import Path
```

2. Localisation du fichier csv 

```python
#Chemin du fichier csv 
fichier_csv=Path.cwd().parent /"donnees/donnees_originales/qualite_service_2025/2025_QoS_Metropole_data_habitations.csv"
#Vérifier que le fichier existe
if fichier_csv.is_file():
    print("Fichier trouvé : ",fichier_csv)
else : 
    raise FileNotFoundError(f"Fichier introuvable : {fichier_csv}")
```

3. Lecture de fichier CSV complet 

```python


donnees_completes = pd.read_csv(
    fichier_csv,
    sep=";",
    encoding="cp1252",
    decimal=",",
    dtype={"insee_com": "string"},
    low_memory=False
)

print("Lecture du fichier terminée.")

```

**) les paramètres : 


--) sep=";" : veut dire que les colonnes dans ce fichier sans séparés par des ';'



--) encoding="cp1252" : c'est pour eviter les erreurs a cause des caractères speciaux comme é , è , ... ect 


--) decimal="," : les chiffres dicimuax sont lus avec des virgules ','


--) dtype={"insee_com": "string"} : dit a pandas de considére la colonne insee_com comme un string pour eviter d'avoir plusieurs type de valeurs pour une seule colonne 


--) low_memory=False : dit a pandas de bien determiner les types des colonnes pendant la lecture du csv en evittant certaine erreurs de detections liées a la lecture par morceaux , cela peut utiliser d'avantage de la memoire

10. Examinons les valeurs manquantes dans l'ensemble du fichier

```python
bilan_manquants=pd.DataFrame({
    "Valeurs manquantes" : donnees_completes.isna().sum(),
    "Pourcentage manquant (%)": donnees_completes.isna().mean() *100
})

bilan_manquants=bilan_manquants.sort_values(by="Pourcentage manquant (%)",ascending=False)

display(bilan_manquants.head(75).round(2))
```

**) pour bilan manquant on a bien indiqué les colonnes de notre dataframe qui va avoir des informations sur les valeurs manquantes et le pourcentage manquants (les colonnes du dframe)



**) isna.sum() : calcule pour chaque colonnes le nombre de valeurs manquantes 


**) isna.mean() *100 : calcule pour chaque colonnes le pourcentage des valeurs manquantes


**)sort_values : permet de trié les lignes de dframe


**)dans les parametres de la methode sort_values on a mis by pourcentage manquants cela veut dire que notre tri concerne la colonne pourcentage manuqants et ascending= false : c'est pour ordre decroissant (par defaut ordre croissant) 



11. Classification des colonnes selon leurs valeurs manquantes (colonnes entierement vide,partiellement et complètes)

```python
total_lignes=len(donnees_completes)
nombre_manquantes=donnees_completes.isna().sum()

colonnes_vides=nombre_manquantes[nombre_manquantes==total_lignes]
colonnes_partielles=nombre_manquantes[(nombre_manquantes > 0) & (nombre_manquantes < total_lignes)]
colonnes_completes=nombre_manquantes[nombre_manquantes==0]

print("Colonnes entièrement vides : ", len(colonnes_vides))
print("Colonnes partiellement renseignés : ", len(colonnes_partielles))
print("Colonnes sans valeurs manquantes : ", len(colonnes_completes))
```

**) nombre_manquantes : c'est la tableau des colonnes et le nombre de valeurs manquantes pour chaque colonne

**)colonnes_vide= va contenir toutes les colonnes qui sont vides (une liste de colonnes vide)
    on trouve ces colonnes vide grace a la condition (nombre_manquantes==total_lignes) qui compare chaque valeur du tableau combres_manquante avec le nombre total de lignes qu'on a de base si c'est bien le cas donc on recupere la colonne et on l'ajoute a notre liste colonnes_vide

**)colonnes_partielles : pareil meme principe avec colonnes_vide juste que on a 2 condition 
    la premiere c'est pour voir si on est pas sur une colonne vide pour bien separés les colonnes vides et partiellement renseignés (parceque la deuxieme condition ne suffit pas vu que une colonne vide peut etre considére comme partiellement renseignés

**)colonnes_completes : pareil 

```python
print("\nColonnes entièrement vides :")
print(colonnes_vides.index.tolist())

print("\nColonnes partiellement renseignés :")
print(colonnes_partielles.index.tolist())

print("\nColonnes sans valeurs manquantes : ")
print(colonnes_completes.index.tolist())
```

**) index : c'est pour recuerper les noms des colonnes et pas les index 

**)tolist() : pour transformer la series pandas en list
