```python
donnees_completes = pd.read_csv(
    fichier,
    sep=";",
    encoding="cp1252",
    decimal=",",
    dtype={"insee_com": "string"}
)
#reconnaitre les nombres decimaux avec la virgule
print("Dimensions :", donnees_completes.shape)

print("Type de la colonne insee_com :")
print(donnees_completes["insee_com"].dtype)

display(donnees_completes["operator"].value_counts(dropna=False))
```

```python
# Afficher les informations générales du dataset
donnees_completes.info()
```

```python
# Compter les valeurs manquantes dans chaque colonne
nombre_manquant = donnees_completes.isna().sum()
#isna() permet de reperer les valeurs manquantes sum() permet de compter ces valeurs 

#Calculer le pourcentage de valeurs manquantes
pourcentage_manquant=donnees_completes.isna().mean() * 100
# isna().mean() calcule la portion des valeurs manquantes (avec des 0 et des 1) *100 indique le pourcentage de cette portion

#Regrouper les resultats dans un DataFrame
resume_manquants = pd.DataFrame({"Nombre de valeurs manquantes ":nombre_manquant,"Pourcentage (%)":pourcentage_manquant})


#Trier les colonnes du plus grand au plus petit pourcentage
resume_manquants=resume_manquants.sort_values(by="Pourcentage (%)",ascending=False)
#sort_values() est une methode pandas qui permet de trier les valeurs dans un ordre précis 
#by = pourcentage c'est pour dire que le tri sera effectuer selon la colonne du pourcentage et ascending=false
#pour avoir le tris du pourcentage plus elevé car par defaut sans ascending ca sera en plus petit (ordre croissant)


#Afficher les 20 colonnes les plus concernées
display(resume_manquants.head(75))
#remaque du resultat: on remarque que beaucoup de colonnes ne possede aucune valeur 


# Objectif de cette exploration de toutes les valeurs manquantes du fichier : examiner les valeurs manquantes dans toutes les colonnes du dataset
# afin d'avoir une vue générale de la qualité des données.
# Cela permet d'identifier les colonnes entièrement ou partiellement vides
# avant de sélectionner les colonnes utiles à notre analyse et de commencer le nettoyage.
```

```python
# Objectif : classer les colonnes selon leurs valeurs manquantes
# afin d'évaluer la qualité globale des données avant le nettoyage.
# IMPORTANT :
# nombre_manquantes est une Series Pandas contenant le nom de chaque colonne
# et son nombre de valeurs manquantes.
#
# Lorsqu'on compare cette Series à un nombre, Pandas effectue automatiquement
# la comparaison pour chaque élément et retourne une Series de booléens (True/False).
#
# Les crochets [condition] permettent de filtrer la Series :
# - True : l'élément est conservé.
# - False : l'élément est ignoré.
#
# Le résultat est une nouvelle Series contenant uniquement les éléments
# qui respectent la condition, avec leur nom et leur nombre de valeurs manquantes.


# Nombre total de lignes
total_lignes = len(donnees_completes)
#on a calculer le nombre total de lignes de notre dataframe (fichier de donnees) avec la fonction len
#qui calcule le nombres de lignes 



# Nombre de valeurs manquantes par colonne
nombre_manquantes = donnees_completes.isna().sum()
# isna() repère les valeurs manquantes (True) et les valeurs présentes (False).
# sum() additionne les True séparément pour chaque colonne.
# Le résultat est une Series Pandas contenant le nom de chaque colonne
# et son nombre de valeurs manquantes. 




# Identifier les trois catégories de colonnes
colonnes_vides = nombre_manquantes[nombre_manquantes == total_lignes]
#on verifie pour la serie pandas des avleurs manquantes combien de colonnes qui sont entirement vide
colonnes_partielles = nombre_manquantes[
    (nombre_manquantes > 0) & (nombre_manquantes < total_lignes)
]

colonnes_completes = nombre_manquantes[nombre_manquantes == 0]

# Afficher les résultats
print("Colonnes entièrement vides :", len(colonnes_vides))
print("Colonnes partiellement renseignées :", len(colonnes_partielles))
print("Colonnes sans valeurs manquantes :", len(colonnes_completes))
```

```python
#Objectif : examiner les colonnes entièrement vides, partiellement renseignées 
#et complètes afin d'identifier les variables utiles à notre étude 
#et de déterminer celles à approfondir pour répondre à notre problématique.

#Afficher les noms des colonnes entierement vides 
print("Colonne entièrement vides :")
print(colonnes_vides.index.tolist())
#index permet de recuperer que le nom des colonnes de notres series pandas qu'on a eu dans la precedante etape
#index correspond aux ettiquettes qui sont dans notre cas colonnes
#tolist() transforme le resultat prec en liste classique de python


# Afficher les noms des colonnes partiellement renseignées
print("\nColonnes partiellement renseignées :")
print(colonnes_partielles.index.tolist())


# Afficher les colonnes entièrement renseignées
print("\nColonnes entièrement renseignées :")
print(colonnes_completes.index.tolist())
```

```python
# Objectif : identifier les protocoles utilisés par l'ARCEP pour évaluer
# différents usages du réseau mobile et connaître le nombre de mesures pour chacun.

# Répartition des différents protocoles de mesure 
display(donnees_completes["protocole"].value_counts(dropna=False))
```

```python
#Notre objectif est de déterminer quels indicateurs sont renseignés pour 
# chaque type de test (protocole).

repartition_champs = donnees_completes.groupby("protocole")[colonnes_partielles.index].count() 
# groupby("protocole") regroupe les observations selon leur protocole.
# count() compte les valeurs non manquantes de chaque colonne
# pour chaque protocole.
# Cela permet d'identifier les indicateurs renseignés selon le type de test.
```

cette etape nous fait comprendre pourquoi certaine colonnes sont partiellement renseignés dans le fichier (car pour chaque protocole par exemple y'a des indicateurs qui utilise et que les autre n'utilisent pas )

```python
#Nombre total d'observations pouur chaque protocole
total_protocole=donnees_completes.groupby("protocole").size()
print(total_protocole)
pourcentage_renseigne=repartition_champs.div(total_protocole,axis=0) *100
display(pourcentage_renseigne.round(2))
```

```python
# Sélectionner les observations du protocole DOWNLOAD
mesures_download = donnees_completes[
    donnees_completes["protocole"] == "DOWNLOAD"
]

# Calculer les statistiques du débit descendant par opérateur
stats_debit_dl = mesures_download.groupby("operator")["bitrate_dl"].describe()

# Afficher les résultats
display(stats_debit_dl.round(2))

#bitrate_dl: c'est le debit descendant autrement dit c'est la vitesse a laquelle le telephone recoit les donnees depuis internet
#objectif : déterminer quel opérateur présente les meilleurs débits descendants selon les mesures réalisées par l'ARCEP.
#Remarque : resultat : Orange possede le meilleur debit decsendant
```

```python
mesures_upload=donnees_completes[donnees_completes["protocole"]=="UPLOAD"]
stats_debit_ul=mesures_upload.groupby("operator")["bitrate_ul"].describe()
display(stats_debit_ul.round(2))
#bitrate_ul : c'est le debit montant c_a_d la vitesse d'envoi des données 
#Remarque : resultat : Orange possede le meilleur debit montant 
```

```python
mesures_web = donnees_completes[
    donnees_completes["protocole"] == "WEB"
]
stats_duree_web=mesures_web.groupby("operator")["acess_duration"].describe()
display(stats_duree_web.round(2))
# Conclusion :
# Orange présente la durée moyenne d'accès aux pages Web la plus faible
# (3,06 secondes), suivi de Bouygues, SFR et Free.
# Orange obtient donc les meilleurs résultats sur cet indicateur.
```

```python
# Sélectionner les observations du protocole STREAM
mesures_stream = donnees_completes[
    donnees_completes["protocole"] == "STREAM"
]
# Calculer les taux de qualité vidéo par opérateur
taux_video = mesures_stream.groupby("operator")[
    ["quality_correct", "quality_perfect"] 
    #ici les indicateur pour stream qui est la lecture de video en continue sans coupure
].mean() * 100 # ici les indicateurs ont été evalués par 0 et 1 donc en binaire et donc on calcule la moyenne de ces 0 et 1

# Afficher les résultats
display(taux_video.round(2))
# Conclusion :
# Orange obtient les meilleurs résultats pour les deux indicateurs
# de qualité vidéo, avec 93,76 % de qualité correcte
# et 92,89 % de qualité parfaite.
# Free obtient les taux les plus faibles.
```

```python
# Sélectionner les observations du protocole ULH
mesures_ulh = donnees_completes[
    donnees_completes["protocole"] == "ULH"
]
# Calculer le taux de réussite des envois par opérateur
taux_upload_ulh = mesures_ulh.groupby("operator")["upload_ok"].mean() * 100

# Afficher les résultats
display(taux_upload_ulh.round(2))

# Conclusion :
# Orange obtient le meilleur taux de réussite des tests ULH avec 92,69 %.
# Free présente le taux le plus faible avec 86,25 %.
# Les résultats montrent des différences de réussite entre les opérateurs.
```
