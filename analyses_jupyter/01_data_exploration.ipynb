```python
import pandas as pd
from pathlib import Path

#chemin du fichier
fichier = Path.cwd().parent / "donnees/donnees_originales/qualite_service_2025/2025_QoS_Metropole_data_habitations.csv"

# Vérifier que le fichier existe
print("Fichier trouvé :", fichier.exists())
#Lecture des 5 premieres lignes 

donnees=pd.read_csv(fichier,sep=";",nrows=5)

#Affichage des informations 
print("Nombre de colonnes : ",len(donnees.columns))
print("\n Noms des colonnes : ")
print(donnees.columns.tolist())
print("\n Apercu des données :")
display(donnees) #affichage des données formatés
```

```python
#lire echantillon de 1000 lignes

echantillon=pd.read_csv(fichier,sep=";",nrows=1000,encoding="cp1252")

#affichage des dimensions de l'echantillon (nbr de lignes et colonnes)

print("Dimensions: ",echantillon.shape)

#informations sur les colonnes

print("\nTypes des colonnes : ")
display(echantillon.dtypes.to_frame("Type"))
#dtypes proprietes de pandas retourne le type de chaque colonne (float int ...)
#to_frame(type) transforme la colonne de données obtenue avec dtypes en data frame ( tebaleau deux dimensions et nomme la colonnes creer type (type de colonne))
#identification des operateurs
print("\nRépartition des opérateurs :")
display(echantillon["operator"].value_counts(dropna=False))
#operateur c'est la colonne de l'operateur telephonique
#value_count() compte le nombre d'occurences de chaque valeur 
#dropna = false compte aussi les valeur manquantes NaN
```

```python
donnees_completes = pd.read_csv(
    fichier,
    sep=";",
    encoding="cp1252",
    decimal=",",
    dtype={"insee_com": "string"}
)
#reconnaitre les nombres decimaux avec la virgule
print("Dimensions :", donnees_completes.shape)

print("Type de la colonne insee_com :")
print(donnees_completes["insee_com"].dtype)

display(donnees_completes["operator"].value_counts(dropna=False))
```

```python
# Afficher les informations générales du dataset
donnees_completes.info()
```

```python
# Compter les valeurs manquantes dans chaque colonne
nombre_manquant = donnees_completes.isna().sum()
#isna() permet de reperer les valeurs manquantes sum() permet de compter ces valeurs 

#Calculer le pourcentage de valeurs manquantes
pourcentage_manquant=donnees_completes.isna().mean() * 100
# isna().mean() calcule la portion des valeurs manquantes (avec des 0 et des 1) *100 indique le pourcentage de cette portion

#Regrouper les resultats dans un DataFrame
resume_manquants = pd.DataFrame({"Nombre de valeurs manquantes ":nombre_manquant,"Pourcentage (%)":pourcentage_manquant})


#Trier les colonnes du plus grand au plus petit pourcentage
resume_manquants=resume_manquants.sort_values(by="Pourcentage (%)",ascending=False)
#sort_values() est une methode pandas qui permet de trier les valeurs dans un ordre précis 
#by = pourcentage c'est pour dire que le tri sera effectuer selon la colonne du pourcentage et ascending=false
#pour avoir le tris du pourcentage plus elevé car par defaut sans ascending ca sera en plus petit (ordre croissant)


#Afficher les 20 colonnes les plus concernées
display(resume_manquants.head(75))
#remaque du resultat: on remarque que beaucoup de colonnes ne possede aucune valeur 


# Objectif de cette exploration de toutes les valeurs manquantes du fichier : examiner les valeurs manquantes dans toutes les colonnes du dataset
# afin d'avoir une vue générale de la qualité des données.
# Cela permet d'identifier les colonnes entièrement ou partiellement vides
# avant de sélectionner les colonnes utiles à notre analyse et de commencer le nettoyage.
```

```python
# Nombre total de lignes
total_lignes = len(donnees_completes)
#on a calculer le nombre total de lignes de notre dataframe (fichier de donnees) avec la fonction len
#qui calcule le nombres de lignes 



# Nombre de valeurs manquantes par colonne
nombre_manquantes = donnees_completes.isna().sum()
#calculons le nombre total de donnees manquantes dans tout le fichier 

print(nombre_manquantes)
# Identifier les trois catégories de colonnes
colonnes_vides = nombre_manquantes[nombre_manquantes == total_lignes]
#on verifie pour 
colonnes_partielles = nombre_manquantes[
    (nombre_manquantes > 0) & (nombre_manquantes < total_lignes)
]

colonnes_completes = nombre_manquantes[nombre_manquantes == 0]

# Afficher les résultats
print("Colonnes entièrement vides :", len(colonnes_vides))
print("Colonnes partiellement renseignées :", len(colonnes_partielles))
print("Colonnes sans valeurs manquantes :", len(colonnes_completes))
```
