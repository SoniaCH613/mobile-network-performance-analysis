```python
import pandas as pd
from pathlib import Path

#chemin du fichier
fichier = Path.cwd().parent / "donnees/donnees_originales/qualite_service_2025/2025_QoS_Metropole_data_habitations.csv"

# Vérifier que le fichier existe
print("Fichier trouvé :", fichier.exists())
#Lecture des 5 premieres lignes 

donnees=pd.read_csv(fichier,sep=";",nrows=5)

#Affichage des informations 
print("Nombre de colonnes : ",len(donnees.columns))
print("\n Noms des colonnes : ")
print(donnees.columns.tolist())
print("\n Apercu des données :")
display(donnees) #affichage des données formatés
```

```python
#lire echantillon de 1000 lignes

echantillon=pd.read_csv(fichier,sep=";",nrows=1000,encoding="cp1252")

#affichage des dimensions de l'echantillon (nbr de lignes et colonnes)

print("Dimensions: ",echantillon.shape)

#informations sur les colonnes

print("\nTypes des colonnes : ")
display(echantillon.dtypes.to_frame("Type"))
#dtypes proprietes de pandas retourne le type de chaque colonne (float int ...)
#to_frame(type) transforme la colonne de données obtenue avec dtypes en data frame ( tebaleau deux dimensions et nomme la colonnes creer type (type de colonne))
#identification des operateurs
print("\nRépartition des opérateurs :")
display(echantillon["operator"].value_counts(dropna=False))
#operateur c'est la colonne de l'operateur telephonique
#value_count() compte le nombre d'occurences de chaque valeur 
#dropna = false compte aussi les valeur manquantes NaN
```

```python
donnees_completes = pd.read_csv(
    fichier,
    sep=";",
    encoding="cp1252",
    decimal=",",
    dtype={"insee_com": "string"}
)
#reconnaitre les nombres decimaux avec la virgule
print("Dimensions :", donnees_completes.shape)

print("Type de la colonne insee_com :")
print(donnees_completes["insee_com"].dtype)

display(donnees_completes["operator"].value_counts(dropna=False))
```
